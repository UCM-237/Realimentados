# Ejercicio 5: Krasovskii-LaSalle

*(Traducción a Python de `ejercicio5.m`, usando `sympy`)*

$$\dot x_1 = x_2, \qquad \dot x_2 = -x_1^3-\gamma x_2$$

```python
import sympy as sp
sp.init_printing()
```

```python
x1, x2, gamma = sp.symbols('x1 x2 gamma', real=True)

f = sp.Matrix([
    x2,
    -x1**3 - gamma*x2
])
f
```

Obtenemos los puntos de equilibrio,

```python
soluciones = sp.solve(f, [x1, x2], dict=True)
soluciones
```

El único punto de equilibrio es el origen.

Definimos $V$, una candidata natural de tipo "energía" para este sistema (energía potencial $x_1^4/4$ más energía cinética $x_2^2/2$).

```python
V = x1**4/4 + x2**2/2
V
```

```python
grv = sp.Matrix([sp.diff(V, x1), sp.diff(V, x2)])
grv
```

```python
vdot = (grv.T * f)[0, 0]
vdot = sp.simplify(vdot)
vdot
```

## Aplicar Krasovskii-LaSalle

$\dot V = -\gamma x_2^2$ es **semidefinida negativa** (se anula en toda la recta $x_2=0$, no solo en el origen), así que el criterio de Lyapunov clásico no basta para concluir estabilidad asintótica directamente.

Aquí es donde entra el **principio de invariancia de LaSalle**: buscamos el mayor conjunto invariante contenido en $\{x : \dot V(x) = 0\} = \{x_2 = 0\}$. Si $x_2\equiv 0$ en algún intervalo de tiempo, entonces también $\dot x_2 \equiv 0$, y sustituyendo en la segunda ecuación, $\dot x_2 = -x_1^3-\gamma\cdot 0 = -x_1^3$, que solo es cero si $x_1=0$. Por tanto, el único conjunto invariante contenido en $\{x_2=0\}$ es el propio origen $\{(0,0)\}$.

Como además $V$ es radialmente no acotada (crece indefinidamente cuando $\|x\|\to\infty$) y definida positiva, LaSalle nos permite concluir que el origen es **globalmente asintóticamente estable** para $\gamma>0$.
