# Modelo del cochecito

Diseñamos el sistema dinámico del cochecillo y lo simulamos para diferentes valores de la señal de control

Ecuaciones del sistema
$$
\dot x = V\cos(\psi)\\
\dot y = V\sin(\psi)\\
\dot \psi = \frac{V}{l}\tan(\phi)
$$

$V = cte$ es la velocidad de avance del coche.  
$l$ es la batalla (distancia entre ejes).  
$\phi$ ángulo de las ruedas con el eje del longitudinal del coche.

```python
import numpy as np
from scipy.integrate import solve_ivp
%matplotlib qt
import matplotlib.pyplot as plt
```

Definimos la ecuación de estado del sistema

```python
def cochecillo(t, x, u, v, l=1.0):
    """
    Modelo de cochecillo (carrito) en el plano XY.
    
    Parámetros:
    t : floatv
        Tiempo actual (no se utiliza en este modelo, pero es necesario para la integración).
    x : array_like
        Estado del sistema [x_pos, y_pos, theta, v], donde:
        - x_pos: posición en X
        - y_pos: posición en Y
        - theta: ángulo de orientación del cochecillo
    u : array_like
        Entrada de control omega, donde:
        - a: aceleración lineal
        - omega: velocidad angular (cambio de orientación)
    parameters:
    v : float
        Velocidad del cochecillo
    l : float, optional
        Distancia entre los ejes del cochecillo (predeterminado: 1.0)
    """
    # Desempaquetar el estado y la entrada de control
    x_pos, y_pos, theta = x
    omega = u(t)

    # Ecuaciones del modelo
    dxdt = v * np.cos(theta)  # Cambio en la posición X
    dydt = v * np.sin(theta)  # Cambio en la posición Y
    dthetadt = v/l*np.tan(omega)  # Cambio en la orientación


    return [dxdt, dydt, dthetadt]
```

```python
v = 10.0  # Velocidad constante del cochecillo
L = 1.0  # Longitud de la batalla del cochecillo
x0 = [0, 0, 0]  # Estado inicial: [x_pos, y_pos, theta]
u = lambda t: 0.1  # Entrada de control ( angulo ruedas constante)
#u = lambda t: 0.1 * np.sin(t)  # Entrada de control como función del tiempo
t_span = (0, 10)  # Intervalo de tiempo para la simulación
# Resolver la ecuación diferencial usando solve_ivp
sol = solve_ivp(cochecillo, t_span, x0, args=(u, v, L), dense_output=True)
```

Nota: Cuando activas dense_output=True, la función te devuelve un objeto especial llamado interpolante polinómico. Este objeto te permite "evaluar" la solución en cualquier instante de tiempo que desees a posteriori, de forma casi instantánea y sin tener que volver a resolver la ecuación diferencial.

Es más eficiente que calcular con un maxstep o un stepeval pequeño.

Vamos a pintar resultados

```python
#graficar la trayectoria del cochecillo
t_eval = np.linspace(t_span[0], t_span[1], 100) #elijo 100 puntos de tiempo para evaluar la solución
x_eval = sol.sol(t_eval) #evaluo la solución en los puntos de tiempo seleccionados
#los pasos anteriores solo son posibles gracias a que se definió dense_output=True
# en la función solve_ivp
plt.figure()
plt.plot(x_eval[0], x_eval[1]) #graficar la trayectoria en el plano XY
plt.xlabel('Posición X')
plt.ylabel('Posición Y')
plt.title('Trayectoria del cochecillo')
```
